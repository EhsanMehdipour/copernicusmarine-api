# Downloading PFT and SST dataset For PS113 Expedition

This notebook guide you to download the satellite dataset necessary for the gap-filling and data fusion of satellite datasets along a corridor track of the RV *Polarster* PS113 expedition. 

This dataset is related to the following studies:
- Mehdipour, E., Xi, H., Barth, A., Alvera-Azcárate, A., Wilhelm, A., and Bracher, A.: Assessment of gap-filling techniques applied to satellite phytoplankton composition products for the Atlantic Ocean, Geosci. Model Dev., 19, 1619–1643, https://doi.org/10.5194/gmd-19-1619-2026, 2026.
- Mehdipour, E., Xi, H., Barth, A., Wilhelm, A., and Bracher, A.: Fusion of Multi-Source Data for Comprehensive Assessment of Phytoplankton Composition, IEEE JSTARS, (Under-Review)

## Prerequisite

Before running this notebook, install the Python environment and authenticate with the Copernicus Marine service as described in `README.md`.

### Import Module

```python
# Copernicu Marine Service module to access and download the data
import copernicusmarine

# Modules to check and visualise the data
import xarray as xr
from matplotlib.colors import LogNorm
import re
```

### Parameters

The spatial and temporal boundaries were defined to encompass the full trajectory of the RV *Polarstern* PS113 expedition together with an extended surrounding observational period for environmental and biogeochemical context analysis. The expedition began on the Patagonian Shelf in the southwestern Atlantic Ocean and concluded in the English Channel. Therefore, the geographical bounds were selected to include the complete cruise pathway across the Atlantic Ocean and adjacent regions relevant to the study.

The temporal window spans three years, from April 25, 2016 to April 25, 2019, thereby including the duration of the expedition as well as the preceding and subsequent environmental conditions captured by the Copernicus Marine products. This extended period enables the investigation of broader temporal variability surrounding the expedition observations.

```python
# Define spatial and temporal boundaries corresponding to the
# RV Polarstern PS113 expedition study region and analysis period

boundaries = {
    "minimum_longitude": -64,
    "maximum_longitude": 3,
    "minimum_latitude": -50,
    "maximum_latitude": 52,
    "start_datetime": "2016-04-25T00:00:00",
    "end_datetime": "2019-04-25T23:59:59",
}
```

## Phytoplankton Functional Types

```python
# Define dataset parameters in a dictionary
dataset_params = {
    "dataset_id": "cmems_obs-oc_glo_bgc-plankton_my_l3-multi-4km_P1D",
    "variables": ["CHL", "DIATO", "DINO", "HAPTO", "GREEN", "PROKAR", "flags"],
    "minimum_longitude": boundaries["minimum_longitude"],
    "maximum_longitude": boundaries["maximum_longitude"],
    "minimum_latitude": boundaries["minimum_latitude"],
    "maximum_latitude": boundaries["maximum_latitude"],
    "start_datetime": boundaries["start_datetime"],
    "end_datetime": boundaries["end_datetime"],
}
```

### 1. Subset

```python
# Generate output filename based on parameters
output_filename = f"{dataset_params['dataset_id']}_{dataset_params['start_datetime'][:10]}_{dataset_params['end_datetime'][:10]}_{dataset_params['minimum_latitude']}_{dataset_params['maximum_latitude']}_{dataset_params['minimum_longitude']}_{dataset_params['maximum_longitude']}.nc"

# Download the subset using the dictionary parameters
copernicusmarine.subset(
    dataset_id=dataset_params["dataset_id"],
    variables=dataset_params["variables"],
    minimum_longitude=dataset_params["minimum_longitude"],
    maximum_longitude=dataset_params["maximum_longitude"],
    minimum_latitude=dataset_params["minimum_latitude"],
    maximum_latitude=dataset_params["maximum_latitude"],
    start_datetime=dataset_params["start_datetime"],
    end_datetime=dataset_params["end_datetime"],
    output_filename=output_filename,
    output_directory=f"data/{dataset_params['dataset_id']}/"
)
```

### 2. open_dataset

```python
ds = copernicusmarine.open_dataset(
    dataset_id=dataset_params["dataset_id"],
    minimum_longitude=dataset_params["minimum_longitude"],
    maximum_longitude=dataset_params["maximum_longitude"],
    minimum_latitude=dataset_params["minimum_latitude"],
    maximum_latitude=dataset_params["maximum_latitude"],
    start_datetime=dataset_params["start_datetime"],
    end_datetime=dataset_params["end_datetime"],
    variables=dataset_params["variables"],
    chunk_size_limit=-1
)
ds
```

```python
## Visualise one day of data
ds.isel(time=5).CHL.plot(robust=True, norm=LogNorm())

# Generate output filename based on parameters
output_filename = f"{dataset_params['dataset_id']}_{dataset_params['start_datetime'][:10]}_{dataset_params['end_datetime'][:10]}_{dataset_params['minimum_latitude']}_{dataset_params['maximum_latitude']}_{dataset_params['minimum_longitude']}_{dataset_params['maximum_longitude']}.nc"

## Save thedata to a new file
ds.to_netcdf(f'data/{dataset_params["dataset_id"]}/{output_filename}')
```

### 3. get (Complete 4 years of global data)

```python
# Define dataset parameters in a dictionary
dataset_id = "cmems_obs-oc_glo_bgc-plankton_my_l3-multi-4km_P1D"
dataset_params = {
    "dataset_id": dataset_id,
    "filter": r"201[6-9]-\d{2}-\d{2}T.*",
    "output_directory": f"./data/{dataset_id}/",

}
# Call the get function to save data
get_result_annualmean = copernicusmarine.get(
    dataset_id=dataset_params["dataset_id"],
    filter=dataset_params["filter"],
    output_directory=dataset_params["output_directory"],
    no_directories=True
)
```
